```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import scanpy as sc
plt.style.use('default')
```

```python
# Load data
df = pd.read_csv('Drugrepurposing/GBM_U251MG_image_algo_5.5_2.txt', sep='\t') # Read data file
```

```python
# Option 1: Scatter plot + kernel density estimation color

sns.set_theme(style="white")
plt.figure(figsize=(8, 8)) 
ax = sns.scatterplot(data=df, x="a_score", y="b_score", hue=None, size=None,
                     palette="coolwarm", s=10, alpha=0.7)
# plt.title("Scatter Plot of es1 vs es2 with Density")
plt.xlabel("a_score of enrichment score")
plt.ylabel("b_score of enrichment score")
```

```python
g = sns.jointplot(
    data=df, x="a_score", y="b_score",
    kind="scatter",
    height=10, ratio=8, marginal_ticks=True,
)
```

```python
# Option2: Scatter Plot + Density Heatmap

plt.figure(figsize=(8, 8))
ax_main = plt.gca()

sns.kdeplot(
    data=df, x="a_score", y="b_score",
    fill=True, cmap="coolwarm",
    thresh=0.05, 
    levels=20, 
    ax=ax_main
)

sns.scatterplot(
    data=df, x="a_score", y="b_score",
    color="grey",
    s=5, alpha=0.3, edgecolor="white", linewidth=0.5,
    ax=ax_main
)
g.set_axis_labels("es1 of enrichment score", "es2 of enrichment score")
# plt.suptitle("Enrichment Scores (es1 vs es2) Distribution", y=1.02) 
plt.tight_layout()
plt.show()
```
